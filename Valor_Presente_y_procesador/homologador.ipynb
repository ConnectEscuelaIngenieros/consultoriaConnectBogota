# Celda 7: comparación Nombre_Capitulo vs capitulo_homologado

```python
import os
import pandas as pd
import re


ruta_base = r"C:\Users\aleja\Documents\Ingenieria Estadistica\Asignaturas2025B\arpro1\Base de Datos ARPRO\Valor_Presente_y_procesador\output"

path_consolidado = os.path.join(ruta_base, "tablaLooker_sin_capitulos_homologados.csv")              # tu csv final
path_candidatos  = os.path.join(ruta_base, "homologacion_capitulos.csv")    # diccionario


# === 1. Cargar data ===
print("[DEBUG] Cargando consolidado y candidatos...")
data_frame_consolidado = pd.read_csv(path_consolidado)
data_frame_candidatos = pd.read_csv(path_candidatos)

# Nos quedamos solo con capítulos
data_frame_capitulos = data_frame_consolidado[
    data_frame_consolidado["Tipo_Fila"] == "CAPITULO"
].copy()

print(f"[DEBUG] Filas de capítulos en consolidado: {len(data_frame_capitulos)}")


# === 2. Deduplicar listas antes del merge ===
data_frame_capitulos_unicos = (
    data_frame_capitulos[["Codigo_Capitulo", "Nombre_Capitulo"]]
    .drop_duplicates()
    .copy()
)

print(f"[DEBUG] Capítulos únicos (Codigo_Capitulo, Nombre_Capitulo): {len(data_frame_capitulos_unicos)}")

data_frame_candidatos_unicos = (
    data_frame_candidatos[["query", "codigo_candidato", "descripcion_candidato"]]
    .drop_duplicates()
    .copy()
)

print(f"[DEBUG] Candidatos únicos (query, codigo_candidato, descripcion_candidato): {len(data_frame_candidatos_unicos)}")


# === 3. Normalizar y construir llaves de merge ===

def normalizar_capitulo(codigo, nombre):
    nombre_limpio = str(nombre).strip().upper()

    # Si el código es NaN, la llave es solo el nombre
    if pd.isna(codigo):
        return nombre_limpio

    codigo_limpio = str(codigo).strip()
    if "." in codigo_limpio:
        codigo_limpio = codigo_limpio.split(".", 1)[0]

    return f"{codigo_limpio} - {nombre_limpio}"



def normalizar_query(query: str) -> str:
    """
    Ejemplo: "10.0 - ENCHAPES" -> "10 - ENCHAPES"
    """
    if query is None:
        return ""

    texto = str(query).strip()
    partes = texto.split("-", 1)
    if len(partes) == 2:
        codigo_parte = partes[0].strip()
        nombre_parte = partes[1].strip().upper()
        # Nos quedamos solo con la parte antes del primer punto (10.0 -> 10)
        codigo_parte = codigo_parte.split(".", 1)[0].strip()
        return f"{codigo_parte} - {nombre_parte}"
    else:
        return texto.upper()


# Llave en consolidado: Codigo_Capitulo + Nombre_Capitulo
data_frame_capitulos_unicos["llave_capitulo"] = data_frame_capitulos_unicos.apply(
    lambda fila: normalizar_capitulo(fila["Codigo_Capitulo"], fila["Nombre_Capitulo"]),
    axis=1,
)

# Llave en diccionario: query normalizada
data_frame_candidatos_unicos["llave_query"] = data_frame_candidatos_unicos["query"].apply(
    normalizar_query
)

print("[DEBUG] Ejemplos de llaves normalizadas (capítulos):")
print(data_frame_capitulos_unicos[["Codigo_Capitulo", "Nombre_Capitulo", "llave_capitulo"]].head())

print("[DEBUG] Ejemplos de llaves normalizadas (candidatos):")
print(data_frame_candidatos_unicos[["query", "llave_query"]].head())


# === 4. Merge solo sobre listas sin duplicados ===

data_frame_merged = data_frame_capitulos_unicos.merge(
    data_frame_candidatos_unicos[["llave_query", "codigo_candidato", "descripcion_candidato"]],
    left_on="llave_capitulo",
    right_on="llave_query",
    how="left",
)

# Construimos columna homologada solo para debug
data_frame_merged["capitulo_homologado"] = data_frame_merged.apply(
    lambda fila: (
        f"{fila['codigo_candidato']} - {fila['descripcion_candidato']}"
        if pd.notna(fila["codigo_candidato"])
        else None
    ),
    axis=1,
)


# === 5. Estadística de cobertura (debug) ===

total_capitulos_unicos = len(data_frame_capitulos_unicos)
capitulos_homologados = data_frame_merged["capitulo_homologado"].notna().sum()
capitulos_no_homologados = total_capitulos_unicos - capitulos_homologados

print("\n===== ESTADÍSTICAS DE HOMOLOGACIÓN (CAPÍTULOS ÚNICOS) =====")
print(f"Total capítulos únicos en consolidado:     {total_capitulos_unicos}")
print(f"Capítulos con homologación encontrada:     {capitulos_homologados}")
print(f"Capítulos SIN homologación (faltantes):    {capitulos_no_homologados}")

if capitulos_no_homologados > 0:
    print("\n[DEBUG] Capítulos SIN homologación (muestra):")
    print(
        data_frame_merged[
            data_frame_merged["capitulo_homologado"].isna()
        ][["Codigo_Capitulo", "Nombre_Capitulo", "llave_capitulo"]]
        .head(20)
    )

print("\n[DEBUG] Ejemplos de capítulos homologados:")
print(
    data_frame_merged[
        data_frame_merged["capitulo_homologado"].notna()
    ][["Codigo_Capitulo", "Nombre_Capitulo", "capitulo_homologado"]]
    .head(20)
)


# === 6. Capítulos sin homologar (detalles) ===

capitulos_sin_homologar = data_frame_merged[
    data_frame_merged["capitulo_homologado"].isna()
].copy()

print(f"[DEBUG] Capítulos sin homologar: {len(capitulos_sin_homologar)}")

print("\n[DEBUG] Muestra de llaves de capítulos sin homologar:")
print(
    capitulos_sin_homologar[
        ["Codigo_Capitulo", "Nombre_Capitulo", "llave_capitulo"]
    ].head(20)
)

print("\n[DEBUG] Muestra de llaves disponibles en homologacion_capitulos:")
print(
    data_frame_candidatos_unicos[["query", "llave_query"]]
    .drop_duplicates()
    .head(20)
)




```

```python
# === HOMOLOGACIÓN FINAL SOBRE TODO EL CONSOLIDADO ===

# 1. Crear la llave en TODO el consolidado
data_frame_consolidado["llave_capitulo"] = data_frame_consolidado.apply(
    lambda fila: normalizar_capitulo(
        fila["Codigo_Capitulo"],
        fila["Nombre_Capitulo"]
    ),
    axis=1,
)

# 2. Merge contra la tabla ya creada data_frame_merged (la tabla de mapeo)
data_frame_consolidado_homologado = data_frame_consolidado.merge(
    data_frame_merged[["llave_capitulo", "codigo_candidato", "descripcion_candidato"]],
    on="llave_capitulo",
    how="left",
)

# 3. Renombrar columnas finales
data_frame_consolidado_homologado = data_frame_consolidado_homologado.rename(
    columns={
        "codigo_candidato": "Codigo_Capitulo_Homologado",
        "descripcion_candidato": "Nombre_Capitulo_Homologado",
    }
)

# 4. Construir columna combinada de capítulo homologado
data_frame_consolidado_homologado["Capitulo_Homologado"] = data_frame_consolidado_homologado.apply(
    lambda fila: (
        f"{fila['Codigo_Capitulo_Homologado']} - {fila['Nombre_Capitulo_Homologado']}"
        if pd.notna(fila["Codigo_Capitulo_Homologado"])
        else None
    ),
    axis=1
)

print("[DEBUG] Listo: columnas nuevas agregadas al consolidado:")
print(
    data_frame_consolidado_homologado[
        ["Codigo_Capitulo_Homologado", "Nombre_Capitulo_Homologado", "Capitulo_Homologado"]
    ].head()
)

# 5. Guardar
path_consolidado_homologado = os.path.join(
    ruta_base, "salida_con_capitulos_homologados.csv"
)
data_frame_consolidado_homologado.to_csv(path_consolidado_homologado, index=False)
print(f"[DEBUG] Archivo guardado: {path_consolidado_homologado}")
```
